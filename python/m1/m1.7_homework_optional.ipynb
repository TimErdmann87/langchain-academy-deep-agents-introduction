```python
from deepagents import create_deep_agent
from langchain.messages import HumanMessage
from langchain_core.runnables import RunnableConfig
from langgraph.checkpoint.memory import MemorySaver
from models import model
```

```python
agent = create_deep_agent(model=model, checkpointer=MemorySaver())

thread: RunnableConfig = {"configurable": {"thread_id": "t_id_a01"}}

user_messages: list[str] = [
    "How can I increase my productivity? Please list 5 different ways that I can implement immediately.",
    "Thank you, which one do you think offers the best productivity enhancements?",
    "Is silence better than certain acustic 'flow state' background music?",
]

for msg in user_messages:
    res = agent.invoke(input={"messages": [HumanMessage(content=msg)]}, config=thread)

    print("\n\nUser:\t", msg)
    print("Agent:\t", res["messages"][-1].content)
```

#### ANALYZING STATE HISTORY

```python
state_history = list(agent.get_state_history(thread))

for i, entry in enumerate(state_history, 1):
    print(f"\nEntry {i}")
    print(
        entry.config["configurable"]["thread_id"],
        " - ",
        entry.config["configurable"]["checkpoint_id"],
    )
```

#### Visualize branching

```python
# Visualize the checkpoint tree: each checkpoint knows its parent, so a checkpoint
# with more than one child is a place where the thread branched.
# NOTE: re-fetch the history, `state_history` above was captured before branching.
full_history = list(agent.get_state_history(thread))


def cp_id(snapshot) -> str:
    return snapshot.config["configurable"]["checkpoint_id"]


children: dict[str | None, list] = {}
for snapshot in full_history:
    parent = snapshot.parent_config
    parent_id = parent["configurable"]["checkpoint_id"] if parent else None
    children.setdefault(parent_id, []).append(snapshot)

for siblings in children.values():
    siblings.sort(key=lambda s: s.created_at)  # oldest branch first


def label(snapshot) -> str:
    msgs = snapshot.values.get("messages", [])
    if msgs:
        text = str(msgs[-1].content).replace("\n", " ")[:40]
        last = f"{msgs[-1].type}: {text!r}"
    else:
        last = "(no messages)"
    return f"step {snapshot.metadata['step']:>2} | {cp_id(snapshot)[-12:]} | {len(msgs)} msgs | {last}"


def render(node, first: str, rest: str, lines: list[str]) -> None:
    prefix = first
    while True:  # follow linear stretches without indenting further
        kids = children.get(cp_id(node), [])
        if len(kids) > 1:
            tag = f"   <== BRANCH POINT ({len(kids)} children)"
        elif not kids:
            tag = "   <== HEAD"
        else:
            tag = ""
        lines.append(f"{prefix}{label(node)}{tag}")
        prefix = rest
        if len(kids) != 1:
            break
        node = kids[0]
    for i, kid in enumerate(kids):
        is_last = i == len(kids) - 1
        render(
            kid,
            rest + ("└─ " if is_last else "├─ "),
            rest + ("   " if is_last else "│  "),
            lines,
        )


lines: list[str] = []
for root in children[None]:
    render(root, "", "", lines)

print(
    f"{len(full_history)} checkpoints, {sum(len(k) > 1 for k in children.values())} branch point(s)\n"
)
print("\n".join(lines))
```
